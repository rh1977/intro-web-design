### 3.2 Comparison & Logical Operators
Booleans evaluate truths and drive logical branching in software applications.

* **Relational Operators:** Equality (`==`), Inequality (`!=`), Greater/Less than (`>`, `<`, `>=`, `<=`).
* **Logical Operators:** Combining conditions using `and`, `or`, and `not`.

<details>
<summary><b>Subsection: Discussion Question 3.2</b></summary>

> **Prompt for Students:** *What is short-circuit evaluation, and how does Python evaluate `False and (10 / 0 == 1)`?*

<details>
<summary><b>Instructor Guide & Answer Key (Click to expand)</b></summary>

* **Target Answer:** Short-circuit evaluation means Python stops evaluating a logical expression as soon as the outcome is guaranteed. In `False and (...)`, since the first operand is `False`, the `and` expression can never be true, so Python skips evaluating `(10 / 0 == 1)` entirely and avoids throwing a `ZeroDivisionError`.
</details>
